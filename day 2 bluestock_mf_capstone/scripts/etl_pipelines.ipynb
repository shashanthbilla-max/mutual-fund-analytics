```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
username = "root"
password = "1234"

host = "localhost"
port = "3306"
```

```python
database = "mutual_funtd_analysis"

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

print("MySQL Connection Successful")
```

```python
scheme_perf_path = r"C:\Users\user\OneDrive\Desktop\bluestock_mf_capstone\notebooks\cleaned_scheme_performance.csv"

nav_history_path = r"C:\Users\user\OneDrive\Desktop\bluestock_mf_capstone\notebooks\cleaned_nav_history.csv"

transactions_path = r"C:\Users\user\OneDrive\Desktop\bluestock_mf_capstone\notebooks\cleaned_investor_transactions.csv"
```

```python
df_scheme = pd.read_csv(scheme_perf_path)

df_nav = pd.read_csv(nav_history_path)

df_transactions = pd.read_csv(transactions_path)

print("\nCSV Files Loaded Successfully")
```

```python
df_scheme.to_sql(
    name="fact_performance",
    con=engine,
    if_exists="replace",
    index=False
)

print("fact_performance loaded")
```

```python
df_nav.to_sql(
    name="fact_nav",
    con=engine,
    if_exists="replace",
    index=False
)

print("fact_nav loaded")
```

```python
df_transactions.to_sql(
    name="fact_transactions",
    con=engine,
    if_exists="replace",
    index=False
)

print("fact_transactions loaded")
```

```python
perf_count = pd.read_sql(
    "SELECT COUNT(*) AS count FROM fact_performance",
    con=engine
)

nav_count = pd.read_sql(
    "SELECT COUNT(*) AS count FROM fact_nav",
    con=engine
)

txn_count = pd.read_sql(
    "SELECT COUNT(*) AS count FROM fact_transactions",
    con=engine
)

print("\n==============================")
print("ROW COUNT VALIDATION")
print("==============================")

print(f"CSV Performance Rows : {len(df_scheme)}")
print(f"MySQL Performance Rows : {perf_count['count'][0]}")

print()

print(f"CSV NAV Rows : {len(df_nav)}")
print(f"MySQL NAV Rows : {nav_count['count'][0]}")

print()

print(f"CSV Transaction Rows : {len(df_transactions)}")
print(f"MySQL Transaction Rows : {txn_count['count'][0]}")

print("\nData Loaded Successfully")
```
